```python
import pandas as pd
import sqlite3
```

```python
plans_df = pd.read_csv("../data/plans.csv")
claims_df = pd.read_csv("../data/claims.csv")
```

```python
# Clean the data
# Drop duplicates
plans_df = plans_df.drop_duplicates()
claims_df = claims_df.drop_duplicates()
```

```python
# Clean the data
#Drop null values
plans_df = plans_df.dropna()
claims_df = claims_df.dropna()
```

```python
# Coerce date filed to datetime
claims_df['date_filed'] = pd.to_datetime(claims_df['date_filed'])
```

```python
claims_df.info()
```

```python
conn = sqlite3.connect("../coverage.db")
```

```python
# Connect plans and claims data to SQLite database
plans_df.to_sql("plans", conn, if_exists="replace", index=False)
claims_df.to_sql("claims", conn, if_exists="replace", index=False)
```
