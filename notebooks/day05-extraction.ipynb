```python
# Extract data from pdfs and docs in sample_docs folder and save it to files inside raw_text folder
```

```python
import pdfplumber
import os
import docx 
import pytesseract
from pdf2image import convert_from_path
```

```python
# Read PDF file and extract text content
text_content = ""
with pdfplumber.open("../sample_docs/benefits.pdf") as pdf:
    for page in pdf.pages:
        text_content += page.extract_text() + "\n"
```

```python
# Preview the first 500 characters of the extracted text
print(text_content[:500])
```

```python
# Save the extracted text to a file in the raw_text folder
os.makedirs("../raw_text", exist_ok=True)
with open("../raw_text/benefits.txt", "w", encoding="utf-8") as f:
    f.write(text_content.strip())
```

```python
# Same extraction process for DOCX files 
# Read docx file and extract text content
doc = docx.Document("../sample_docs/claims_process.docx")
claims_text = "\n".join([p.text for p in doc.paragraphs if p.text.strip()]) 
print(claims_text[:500])
```

```python
# Save to raw_text/claims_process.txt
with open("../raw_text/claims_process.txt", "w", encoding="utf-8") as f:
    f.write(claims_text.strip())
```

```python
import os
import pytesseract
from pdf2image import convert_from_path

# 1. Convert PDF pages to images
pdf_path = "../sample_docs/enrollment_scan.pdf"
images = convert_from_path(pdf_path)

# 2. Extract text using Tesseract OCR
ocr_text = ""
for page_num, img in enumerate(images, start=1):
    extracted = pytesseract.image_to_string(img)
    ocr_text += f"--- Page {page_num} ---\n" + extracted + "\n\n"

# 3. Preview extracted text
print("--- OCR PREVIEW ---")
print(ocr_text[:300])

# 4. Save output to raw_text/enrollment.txt
os.makedirs("../raw_text", exist_ok=True)
with open("../raw_text/enrollment.txt", "w", encoding="utf-8") as f:
    f.write(ocr_text.strip())

print("\nSuccessfully saved to raw_text/enrollment.txt!")
```

```python
# Using BeautifulSoup to extract text from a website
import requests
from bs4 import BeautifulSoup
```

```python
url = "http://books.toscrape.com"
response = requests.get(url)

if response.ok:
    soup = BeautifulSoup(response.text, "html.parser")
    # Extract all book titles
    book_titles = [h3.a["title"] for h3 in soup.find_all("h3")]
    print("Extracted Book Titles:")
    for title in book_titles:
        print(title)
```
